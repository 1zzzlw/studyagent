```python
import chromadb

client = chromadb.EphemeralClient()
collection = client.get_or_create_collection("ephemeral-client-demo")

# 如果在这里不写embeddings的话，chroma就会使用自己内置的向量大模型，对文本进行向量化
collection.add(
    ids=["chroma-1", "python-1"],
    documents=["Chroma 是一个向量数据库", "Python 是一种编程语言"],
)

result = collection.query(
    query_texts=["python 是什么"],
    n_results=1,
)

print(result)
print(result["ids"][0])
print(result["documents"][0])
print(result["distances"][0])
```
