1. chromadb 的初始化

```python
from pathlib import Path

import chromadb

# 获取在该系统下的绝对路径
db_path = Path("./storage/chroma").resolve()
print("向量数据库的目录是", db_path, type(db_path))

# 这里的路径参数，最好使用绝对路径
client = chromadb.PersistentClient(db_path)
# 这里其实就相当于创建一个表
collection = client.get_or_create_collection("persistent-client-demo")

# 还是使用 chroma 的内置向量大模型
collection.add(
    ids=["chroma-1", "python-1"],
    documents=["Chroma 是一个向量数据库", "Python 是一种编程语言"],
)

result = collection.query(
    query_texts=["向量数据库是什么？", "Python 是什么？"],
    n_results=1
)

# print(result)
print(result["ids"])
print(result["documents"])
print(result["distances"])
```
