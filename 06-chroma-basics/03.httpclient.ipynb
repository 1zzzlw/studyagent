```python
import chromadb

client = chromadb.HttpClient(
    host="localhost",
    port=8000,
    ssl=False,
)

print("服务端心跳：", client.heartbeat())

collection = client.get_or_create_collection("http-client-demo")

collection.add(
    ids=["chroma-1", "python-1"],
    documents=["Chroma 是一个向量数据库", "Python 是一种编程语言"],
)

result = collection.query(
    query_texts=["pyhon是什么"],
    n_results=1,
)

print(result["ids"][0])
print(result["documents"][0])
```
