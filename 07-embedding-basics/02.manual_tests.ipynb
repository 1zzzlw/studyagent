```python
from embedding_lab.service import EmbeddingLabService

service = EmbeddingLabService()
print(type(service).__name__)
print("知识库目录：", service.settings.knowledge_dir)

```

```python
documents, chunks = service.build_chunks()

assert documents, "应该至少扫描到一份 Markdown"
assert chunks, "扫描到文档后应该生成文本块"
# all() 规则：全部为 True，才返回 True；只要有一个 False，直接返回 False
assert all(chunk.id for chunk in chunks)
assert all(chunk.text for chunk in chunks)
assert all("source" in chunk.metadata for chunk in chunks)
assert all("heading" in chunk.metadata for chunk in chunks)

print("文档数量：", len(documents))
print("文本块数量：", len(chunks))
```

```python
first = chunks[0]

print("第一个块 ID：", first.id)
print("第一个块来源：", first.metadata["source"])
print("第一个块标题：", first.metadata["heading"])
print("第一个块正文预览：")
print(first.text[:300])
```
